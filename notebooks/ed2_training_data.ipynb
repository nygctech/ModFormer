# Extended Data Figure 2. Training data and held-out predictions

**Training data, fine-tuning, and held-out editing predictions.** Thin lines are 95% percentile-bootstrap intervals. **a,** Datasets, sizes, and train/validation/test splits. Filled circles mark training uses. Alu and invertebrate duplexes contribute to fine-tuning. The example tile of a pretraining window is not reproduced here. **b,** One-epoch joint fine-tuning over eight sources; bar length, sampling share; dots, the prediction heads each source trains. The diagram of the fine-tuned network is not reproduced here. **c,d,** Editing-site AUC (**c**) and rate Spearman rho (**d**) on held-out human REDIportal and HEK293FT datasets. Bars, zero-shot; squares, fine-tuned; dotted lines, ModFormer's zero-shot value. REDIportal lacks measured rates. **e,** Mean predicted versus measured editing rates for 11,915 adenosines in 90 Alu duplexes. Each point is the mean of one bin of predicted rate: nine deciles, then the 90th-95th percentile, the 95th-99th percentile, and the top 1%; diagonal, equal means. Group means do not measure per-site accuracy. **f,** Editing-site AUC and rate Spearman rho in all 90 Alu duplexes, 69 sharing a 128-nt window with the pretraining tiles and 21 without. Red dot, zero-shot; open square, one-epoch fine-tune; no line joins them. Zero-shot, without versus with overlap: AUC, 0.806 versus 0.822; rho, 0.506 versus 0.504.

Run this notebook from the `notebooks` folder. The tables are read from the folder in `MODFORMER_FIGURE_DATA` when it is set, and otherwise from the `figure-data` folder of the ModFormer cache (`MODFORMER_HOME`), where they are fetched on first use. The source tables are in the ModFormer data record, doi 10.5281/zenodo.24300820; the model weights are in doi 10.5281/zenodo.24300574. This notebook uses the tables `ed2a`, `ed2b`, `ed2cd`, `ed2e` and `ed2f`; the record's `README.md` describes each table, and `figure_source_data/SHA256SUMS.txt` inside `figure_source_data.zip` lists their checksums.

```
import _figures as fg
import _extended as ed

tables = {name: fg.load_table(name) for name in ed.ED2_TABLES}
{name: t.shape for name, t in tables.items()}
```

```
fig, digest = ed.ed2(tables)
```

```
import json
import os
from pathlib import Path

out = Path(os.environ.get("MODFORMER_FIGURE_OUT", "figures-out"))
out.mkdir(parents=True, exist_ok=True)
fig.savefig(out / "ed2.png", dpi=200, bbox_inches="tight")
fig.savefig(out / "ed2.pdf", bbox_inches="tight")
(out / "ed2_digest.json").write_text(json.dumps(digest, indent=1) + "\n")
print("saved", sorted(p.name for p in out.glob("ed2*")))
```
