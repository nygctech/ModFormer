# Extended Data Figure 3. Library screen

**Editing-site prioritization across mutation distances and cellular contexts.** Held-out library constructs; red, zero-shot ModFormer; blue, library-trained model. Top and bottom picks are the sites with greatest predicted editing gain and loss. **a,** Distance from selected sites to the position paired with the nearest mutation (8-nt bins); gray, all candidate sites. **b,** Fraction of constructs with observed editing changes in the predicted direction across five cellular conditions. Dark/light, top/bottom picks; navy, a top-pick fraction below the random baseline; gray, random baseline. The last two conditions are in ADAR1-knockout cells. The first four conditions use library-trained predictions; ADAR1 overexpression uses zero-shot predictions, so enzyme contexts are not compared using identical checkpoints. **c,** Fraction of top picks showing increased editing with one pick per distance band; gray, a random site from the same band. **d,** Fraction showing increased editing versus number of picks per construct; 95% bootstrap intervals and matched random baseline. **e,** Fraction of ranked sites with increased editing versus total sites tested; gray, shuffled predictions. **f,** Recovery of the 50 largest measured gains versus sites tested; gray, shuffled predictions. Panels a, c-f show HEK293T and MCF7. Within 10 nt of the mutation-paired position, the difference in top-pick gain fraction (library-trained minus zero-shot; 95% intervals) was -0.001 (-0.027 to +0.025) in HEK293T, +0.018 (-0.007 to +0.044) with interferon, -0.024 (-0.050 to +0.003) in MCF7, and +0.030 (+0.001 to +0.060) with ADAR2 overexpression.

Run this notebook from the `notebooks` folder. The tables are read from the folder in `MODFORMER_FIGURE_DATA` when it is set, and otherwise from the `figure-data` folder of the ModFormer cache (`MODFORMER_HOME`), where they are fetched on first use. The source tables are in the ModFormer data record, doi 10.5281/zenodo.24300820; the model weights are in doi 10.5281/zenodo.24300574. This notebook uses the tables `ed3a` to `ed3f` and `ed3_top_pick_difference`; the record's `README.md` describes each table, and `figure_source_data/SHA256SUMS.txt` inside `figure_source_data.zip` lists their checksums.
