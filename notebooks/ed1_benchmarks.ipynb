# Extended Data Figure 1. Benchmarks against published predictors

**ModFormer performance against published editing-site predictors.** **a,** Table of method capabilities (not drawn here). **b,** Editing-site AUC across Alu (729 sites), InoDREAM (17,956), invertebrate (1,678), and mouse Neuro-2a/brain (21,667) benchmarks. Gray shading, species absent from model training; rings, evaluation on a method's training-data source; dotted line, chance. ModFormer, AdarEdit, and InoDREAM have 95% bootstrap intervals. Cohorts and labels differ from Figure 2a. **c,** Minimum, maximum, and unweighted mean AUC across benchmarks; methods ordered by mean. **d,** Spearman rho with measured editing rates at edited sites in Alu (1,683), HEK293T (2,461), human brain (775), and mouse Neuro-2a/brain (276,924). Within each method's row, the sets run top to bottom in this order; thin lines, 95% intervals.

Run this notebook from the `notebooks` folder. The tables are read from the folder in `MODFORMER_FIGURE_DATA` when it is set, and otherwise from the `figure-data` folder of the ModFormer cache (`MODFORMER_HOME`), where they are fetched on first use. The source tables are in the ModFormer data record, doi 10.5281/zenodo.24300820; the model weights are in doi 10.5281/zenodo.24300574. This notebook uses the tables `ed1b`, `ed1c` and `ed1d`; the record's `README.md` describes each table, and `figure_source_data/SHA256SUMS.txt` inside `figure_source_data.zip` lists their checksums.

```
import _figures as fg
import _extended as ed

tables = {name: fg.load_table(name) for name in ed.ED1_TABLES}
{name: t.shape for name, t in tables.items()}
```

```
fig, digest = ed.ed1(tables)
```

```
import json
import os
from pathlib import Path

out = Path(os.environ.get("MODFORMER_FIGURE_OUT", "figures-out"))
out.mkdir(parents=True, exist_ok=True)
fig.savefig(out / "ed1.png", dpi=200, bbox_inches="tight")
fig.savefig(out / "ed1.pdf", bbox_inches="tight")
(out / "ed1_digest.json").write_text(json.dumps(digest, indent=1) + "\n")
print("saved", sorted(p.name for p in out.glob("ed1*")))
```
