# Figure 3. Substrate selection

**ModFormer prioritizes active editing sites and responds to ADAR-associated sequence and pairing features.** **a,** Recovery of highly edited Alu sites (at least 10% measured editing) versus the fraction of candidates screened, for ModFormer (red) and a language-model-only baseline (slate); dashed, zero-shot; solid, fine-tuned; bands span the two states; the dotted vertical line marks the fraction of sites that are highly edited. **b,** Sequence and opposing-base enrichment among high-scoring mouse REDIportal adenosines; letter height is log2 fold enrichment over all evaluated adenosines (scale bar, 1- to 2-fold), and the edited A is drawn at the height of the tallest column. **c,** Predicted change in editing logit after creating or removing mismatches opposite the adenosine (A:C is a mismatch, A·U a pair); nearby pairing-preserving substitutions are controls, and the gray box is the control margin; n is the number of sequence families and bars are confidence intervals. The perturbations are computational.

Run this notebook from the `notebooks` folder. The tables are read from the folder in `MODFORMER_FIGURE_DATA` when it is set, and otherwise from the `figure-data` folder of the ModFormer cache (`MODFORMER_HOME`), where they are fetched on first use. The source tables are in the ModFormer data record, doi 10.5281/zenodo.24300820; the model weights are in doi 10.5281/zenodo.24300574.
