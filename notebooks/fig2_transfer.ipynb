# Figure 2. Transfer across species and substrates

**ModFormer predicts editing-site activity and editing rate across RNA substrates and species.** **a,** Editing-site classification AUC on five test sets, zero-shot (bars) and after one epoch of fine-tuning (squares). Dotted lines mark ModFormer's zero-shot value; whiskers (zero-shot) and pale bands (fine-tuned) are 95% bootstrap intervals. **b,** Spearman correlation between predicted and measured editing rates, including measured zeros; marks as in a; REDIportal has no measured rates.

Run this notebook from the `notebooks` folder. The tables are read from the folder in `MODFORMER_FIGURE_DATA` when it is set, and otherwise from the `figure-data` folder of the ModFormer cache (`MODFORMER_HOME`), where they are fetched on first use. The source tables are in the ModFormer data record, doi 10.5281/zenodo.24300820; the model weights are in doi 10.5281/zenodo.24300574.

```
import _figures as fg

table = fg.load_table("fig2_cells.csv")
table.shape
```

```
fig, digest = fg.figure2(table)
```

```
import json
import os
from pathlib import Path

out = Path(os.environ.get("MODFORMER_FIGURE_OUT", "figures-out"))
out.mkdir(parents=True, exist_ok=True)
fig.savefig(out / "figure2.png", dpi=200, bbox_inches="tight")
fig.savefig(out / "figure2.pdf", bbox_inches="tight")
(out / "figure2_digest.json").write_text(json.dumps(digest, indent=1) + "\n")
print("saved", sorted(p.name for p in out.glob("figure2*")))
```
