# Figure 4. Variant effects

**ModFormer predicts mutation effects and prioritizes responsive editing sites.** **b,** Editing-change correlation versus the number of library constructs used for training; filled points exceed the zero-shot value (red line) in a paired comparison, and the dotted line marks the budget used in c. **c,** Editing-change correlation for four models in five cellular conditions; the joint fine-tune and library-trained models were not scored in ADAR1-KO + ADAR1-OE. **d,** Share of constructs whose zero-shot highest-gain or highest-loss site changed in the expected direction; gray boxes, the range for a randomly chosen site. **e,** Share of ranked sites with measured editing gains versus the number tested; the gray band is shuffled predictions. **f,** Correlation between predicted and measured editing changes for three endogenous substrates. Whiskers in c and f are confidence intervals.

Panel a of the paper's Figure 4 combines a drawing of the mutated duplexes with two data tracks, the number of variants by mutation type and the per-position correlation between predicted and measured editing change. It is not reproduced here, so this notebook draws panels b to f only.

Run this notebook from the `notebooks` folder. The tables are read from the folder in `MODFORMER_FIGURE_DATA` when it is set, and otherwise from the `figure-data` folder of the ModFormer cache (`MODFORMER_HOME`), where they are fetched on first use. The source tables are in the ModFormer data record, doi 10.5281/zenodo.24300820; the model weights are in doi 10.5281/zenodo.24300574.
